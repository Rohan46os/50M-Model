```python
import torch.nn as nn
import torch
```

```python
class ElmanRnn(nn.Module):
  def __init__(self, input_size, hidden_size, output_size):
    super().__init__()
    self.input_size = input_size
    self.hidden_size = hidden_size
    self.output_size = output_size

    self.input_to_hidden = nn.Linear(input_size, hidden_size)
    self.context_to_hidden = nn.Linear(hidden_size, hidden_size)
    self.hidden_to_output = nn.Linear(hidden_size, output_size)
    self.activation  = nn.Tanh()


  def forward(self, x_val, previous_hidden_state):
    processed_input = self.input_to_hidden(x_val)
    previous_hidden_state = self.context_to_hidden(previous_hidden_state)
    hidden_state = self.activation(processed_input + previous_hidden_state)
    output = self.hidden_to_output(hidden_state)
    return output, hidden_state

```
